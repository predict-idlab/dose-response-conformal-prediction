```python
import pandas as pd
import numpy as np
```

```python
amount = 250
current_len = 0 
while current_len != amount:
    generated_age_list = 100-np.random.lognormal(3.5,0.6,270)
    final_generated_age_list = np.round(generated_age_list[generated_age_list > 18])
    current_len = len(final_generated_age_list)
```

```python
gender_list = 1+np.random.binomial(1,0.3,250)
```

```python
len(gender_list)
```

```python
weight_list = np.round((gender_list-1)*(-5)+np.random.normal(75,10,250))
```

```python
height_list = np.round((np.random.normal(178,8,250))*(gender_list == 1) + (np.random.normal(164,7,250))*(gender_list == 2))
```

```python
amicas_db_df = pd.DataFrame(np.array([final_generated_age_list,height_list,weight_list,gender_list]).transpose(),columns=["Age","height","weights","sex"])
```

```python
amicas_db_df.to_csv("all_data_amicas.csv",index=False)
```
